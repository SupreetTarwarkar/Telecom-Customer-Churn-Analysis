# Exploratory Data Analysis of Telecom Customer Churn


## Problem Statement
The telecom industry experiences high customer churn. This analysis identifies patterns and factors influencing churn across customer demographics, services, billing details, contracts, and account information to support customer-retention strategies.


## Objectives
- Identify the overall churn rate and compare it with active customers.
- Analyze demographics, services, billing, and account details associated with churn.
- Uncover patterns and trends using exploratory data analysis.
- Provide actionable retention insights.


## Import Libraries


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

## Load Data


```python
data = pd.read_csv('telecom_churn_data.csv')
data.head()
```

## Basic Data Checks


```python
print('Shape:', data.shape)
print('Duplicates:', data.duplicated().sum())
data.info()
```

## Data Cleaning


```python
data.replace({' ': np.nan}, inplace=True)
data['Total_Charges'] = pd.to_numeric(data['TotalCharges'], errors='coerce')
data.drop('TotalCharges', axis=1, inplace=True)
print(data.isnull().sum())
```

The original analysis identified 11 missing values in Total Charges. Because the distribution is right-skewed, the missing values are replaced with the median.


```python
data['Total_Charges'] = data['Total_Charges'].fillna(data['Total_Charges'].median())
print('Remaining missing values:', data.isnull().sum().sum())
```

## Univariate Analysis


```python
plt.figure(figsize=(6,4))
sns.countplot(data=data, x='Contract', palette='plasma')
plt.title('Contract Type Distribution')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=data, x='InternetService', palette='plasma')
plt.title('Internet Service Distribution')
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.histplot(data['tenure'], bins=30, kde=True)
plt.title('Customer Tenure Distribution')
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.histplot(data['MonthlyCharges'], bins=30, kde=True)
plt.title('Monthly Charges Distribution')
plt.show()
```

## Churn Analysis


```python
plt.figure(figsize=(6,4))
sns.countplot(data=data, x='Contract', hue='Churn', palette='plasma')
plt.title('Churn vs Contract Type')
plt.show()
```

**Insight:** Month-to-Month customers show the strongest churn concentration, while One-Year and Two-Year contract customers show much lower churn.


```python
plt.figure(figsize=(6,4))
sns.countplot(data=data, x='InternetService', hue='Churn', palette='plasma')
plt.title('Churn vs Internet Service')
plt.show()
```

**Insight:** Fiber Optic customers show the highest churn levels among the internet-service groups analyzed. Fiber Optic also represents about 44% of customers in the dataset.


```python
plt.figure(figsize=(10,4))
sns.countplot(data=data, x='PaymentMethod', hue='Churn', palette='plasma')
plt.xticks(rotation=20)
plt.title('Churn vs Payment Method')
plt.show()
```

**Insight:** Electronic Check customers show the highest churn levels among the payment methods analyzed.


```python
plt.figure(figsize=(6,4))
sns.boxplot(data=data, x='Churn', y='MonthlyCharges', palette='plasma')
plt.title('Monthly Charges vs Churn')
plt.show()
```

**Insight:** Customers who churned tend to have higher Monthly Charges than customers who stayed.


```python
plt.figure(figsize=(6,4))
sns.countplot(data=data, x='Dependents', hue='Churn', palette='plasma')
plt.title('Churn vs Dependents')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=data, x='TechSupport', hue='Churn', palette='plasma')
plt.title('Churn vs Tech Support')
plt.show()
```

**Insight:** Customers without Tech Support show higher churn levels than customers with Tech Support.


## Overall Churn Rate


```python
count = data['Churn'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(count, labels=count.index, autopct='%1.1f%%', startangle=90, wedgeprops={'width':0.6})
plt.title('Churn Distribution')
plt.show()
```

**Insight:** 26.5% of customers churned while 73.5% stayed, indicating a meaningful retention challenge.


## Internet Service Share


```python
count = data['InternetService'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(count, labels=count.index, autopct='%1.1f%%', startangle=90)
plt.title('Internet Service Distribution')
plt.show()
```

**Insight:** Fiber Optic accounts for about 44% of customers, DSL for 34.4%, and 21.7% have no internet service.


## Conclusion
The analysis highlights contract type, internet service, payment method, monthly charges, and support services as important churn-related dimensions. Retention efforts can prioritize Month-to-Month customers, Fiber Optic users, Electronic Check users, and customers without Tech Support.
